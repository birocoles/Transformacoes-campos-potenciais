## Step 1) Load data and set main field

### Input

```python
# Load the data file
observed_data = np.loadtxt("data.txt")
```

```python
# Set the main field inclination (degrees)
I0 = -30

# Set the main field declination (degrees)
D0 = 10

print('Main field inc = {}° dec = {}°'.format(I0, D0))
```

### Auxiliary information

```python
# Set the characteristic data spacing (in meters)
# In the case of aeromagnetic surveys, this parameter is average flight line spacing
data_spacing = 400.
```

```python
# Compute the unit vector defining the main field direction
F_hat = utils.unit_vector(inc=I0, dec=D0)
```

```python
# Compute the study area limits
# xmin, xmax, ymin, ymax
study_area = [
    np.min(observed_data[:,0]), 
    np.max(observed_data[:,0]), 
    np.min(observed_data[:,1]), 
    np.max(observed_data[:,1])
]
```

## Step 2) Total-field anomaly inversion

```python
# Compute the constant vertical coordinate of the equivalent layer
# Median of the data vertical coordinate + characteristic data spacing
z_layer = np.median(observed_data[:,2]) + data_spacing
```

```python
print(f"Equivalent-layer vertical coordinate = {z_layer:>.2f} m")
```

```python
# Compute a dictionary with data coordinates
coordinates = {
    'x':observed_data[:,0],
    'y':observed_data[:,1],
    'z':observed_data[:,2]
}
```

```python
# compute the SEDM
R2 = inverse_distance.sedm(
    data_points = coordinates, 
    source_points = z_layer
)
```

```python
# compute the first directional derivative along the main field
Grad = inverse_distance.grad(
    data_points = coordinates, 
    source_points = z_layer,
    SEDM = R2,
    components=['x','y','z']
)
```

```python
# estimate the equivalent layer
bz_convergence, bz_layer = eqlayer.method_CGLS(
    sensitivity_matrices=[F_hat[0]*Grad['x']+F_hat[1]*Grad['y']+F_hat[2]*Grad['z']],
    data_vectors=[observed_data[:,3]],
    epsilon=1e-3,
    ITMAX=100,
    p0=None
)
```

```python
# Plot the estimated equivalent layer
bound = 1e-5*np.max(np.abs(bz_layer))

fig, ax = plt.subplots()

ax.axis('scaled')
im = ax.scatter(
        x=coordinates['y']*0.001, y=coordinates['x']*0.001, 
        c=1e-5*bz_layer, s=10,
        cmap='seismic', vmin=-bound, vmax=bound
    )
ax_divider = make_axes_locatable(ax)
cax = ax_divider.append_axes("right", size="5%", pad="2%")
cb = fig.colorbar(im, cax=cax)
cb.ax.tick_params(labelsize=14)
cb.ax.set_xlabel('$\\times 10^{5}$', fontsize=16, loc='left')
ax.set_ylim(0.001*study_area[0], 0.001*study_area[1])
ax.set_xlim(0.001*study_area[2], 0.001*study_area[3])
ax.tick_params(axis='x', labelsize=16)
ax.tick_params(axis='y', labelsize=16)
ax.grid()
ax.set_ylabel('x (km)', fontsize=18)
ax.set_xlabel('y (km)', fontsize=18)

plt.show()
```

```python
del fig, ax, im
gc.collect()
```

```python
# Convergence curve of the iterative total-field anomaly inversion
fig, ax = plt.subplots(figsize=(8,4))

iterations = [i for i in range(len(bz_convergence))]
ax.plot(
    iterations, bz_convergence, linestyle='-', c='k', alpha=1, markersize=4
)
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)
ax.set_xlabel('Iterations', fontsize=16)
ax.set_ylabel('Data fit', fontsize=16)
ax.grid(True)
plt.tight_layout()

plt.show()
```

```python
del fig, ax
gc.collect()
```

### Compute the data fit

```python
# magnetic induction field component x
predicted_bx = Grad['x']@bz_layer

# magnetic induction field component y
predicted_by = Grad['y']@bz_layer

# magnetic induction field component z
predicted_bz = Grad['z']@bz_layer
```

```python
del Grad
gc.collect()
```

```python
# Compute the predicted data (approximated total-field anomaly)
predicted_data = F_hat[0]*predicted_bx + F_hat[1]*predicted_by + F_hat[2]*predicted_bz
```

```python
del predicted_bx, predicted_by, predicted_bz
gc.collect()
```

```python
# Compute the residuals
residuals = predicted_data - observed_data[:,3]
```

```python
# Plot the data fit
fig = plt.figure(
    layout= 'tight', 
    figsize=(15,15)
)
mosaic = fig.subplot_mosaic(
    mosaic = '''
             abc
             ''',
    sharex=True, sharey=True
)

labels = [['(a)', '(b)', '(c)']]

mosaic_elements = [['a', 'b', 'c']]

for (mosaic_row, labels_row) in zip(
    mosaic_elements, 
    labels
):
    data_list = [
        predicted_data, 
        observed_data[:,3],
        residuals
    ]
    bounds_list = 2*[np.max(np.abs(observed_data[:,3]))] + [0.8*np.max(np.abs(residuals))]
    titles = ['ACT calculada', 'ACT simulada', 'resíduos']
    
    for mosaic_element, data, bound, label, title in zip(mosaic_row, data_list, bounds_list, labels_row, titles):
    
        ax = mosaic[mosaic_element]
        ax.axis('scaled')
        im = ax.scatter(
            x=coordinates['y']*0.001, y=coordinates['x']*0.001, 
            c=data, s=10,
            cmap='seismic', vmin=-bound, vmax=bound
        )
        ax_divider = make_axes_locatable(ax)
        cax = ax_divider.append_axes("right", size="5%", pad="2%")
        cb = fig.colorbar(im, cax=cax)
        cb.ax.tick_params(labelsize=14)
        cb.ax.set_xlabel('nT', fontsize=16, loc='left')
        ax.set_ylim(0.001*study_area[0], 0.001*study_area[1])
        ax.set_xlim(0.001*study_area[2], 0.001*study_area[3])
        ax.tick_params(axis='x', labelsize=16)
        ax.tick_params(axis='y', labelsize=16)
        ax.grid()
        ax.annotate(label, xy=(0.02, 0.93), xycoords='axes fraction', fontsize=20)
        ax.annotate(title, xy=(0.02, 0.03), xycoords='axes fraction', fontsize=14)
    
for element in ['a']:
    mosaic[element].set_ylabel('x (km)', fontsize=18)
for element in ['a','b', 'c']:
    mosaic[element].set_xlabel('y (km)', fontsize=18)

plt.savefig('residuos_camada_1.png', dpi=300)

plt.show()
```

```python
del fig, ax, data, mosaic, data_list
gc.collect()
```

## Step 3) Total-magnetization direction

```python
z_layer2 = z_layer - 0.05*data_spacing
```

```python
coordinates['z'] = np.broadcast_to(z_layer2, observed_data.shape[0])
```

```python
# compute the SEDM
R2 = inverse_distance.sedm(
    data_points = coordinates, 
    source_points = z_layer
)
```

```python
# compute the first directional derivative along the main field
Grad = inverse_distance.grad(
    data_points = coordinates, 
    source_points = z_layer,
    SEDM = R2,
    components=['x','y','z']
)
```

```python
# magnetic induction field component x
predicted_bx = Grad['x']@bz_layer

# magnetic induction field component y
predicted_by = Grad['y']@bz_layer

# magnetic induction field component z
predicted_bz = Grad['z']@bz_layer
```

```python
b_amplitude = np.sqrt(predicted_bx**2 + predicted_by**2 + predicted_bz**2)
```

```python
del predicted_bx, predicted_by, predicted_bz, R2, Grad
gc.collect()
```

```python
coordinates['z'] = observed_data[:,2]
```

```python
# Plot the estimated b amplitude
bound = np.max(np.abs(b_amplitude))

fig, ax = plt.subplots()

ax.axis('scaled')
im = ax.scatter(
        x=coordinates['y']*0.001, y=coordinates['x']*0.001, 
        c=b_amplitude, s=10,
        cmap='seismic', vmin=-bound, vmax=bound
    )
ax_divider = make_axes_locatable(ax)
cax = ax_divider.append_axes("right", size="5%", pad="2%")
cb = fig.colorbar(im, cax=cax)
cb.ax.tick_params(labelsize=14)
cb.ax.set_xlabel('$\\times 10^{5}$', fontsize=16, loc='left')
ax.set_ylim(0.001*study_area[0], 0.001*study_area[1])
ax.set_xlim(0.001*study_area[2], 0.001*study_area[3])
ax.tick_params(axis='x', labelsize=16)
ax.tick_params(axis='y', labelsize=16)
ax.grid()
ax.set_ylabel('x (km)', fontsize=18)
ax.set_xlabel('y (km)', fontsize=18)

plt.show()
```

```python
del fig, ax, im
gc.collect()
```

```python
# compute the SEDM
R2 = inverse_distance.sedm(
    data_points = coordinates, 
    source_points = z_layer2
)
```

```python
# compute the second derivatives
Tensor = inverse_distance.grad_tensor(
    data_points = coordinates, 
    source_points = z_layer2,
    SEDM = R2,
    components=['xx', 'xy', 'xz', 'yy', 'yz', 'zz']
)
```

```python
del R2
gc.collect()
```

```python
A = np.vstack([
    (F_hat[0]*Tensor['xx'] + F_hat[1]*Tensor['xy'] + F_hat[2]*Tensor['xz'])@b_amplitude,
    (F_hat[0]*Tensor['xy'] + F_hat[1]*Tensor['yy'] + F_hat[2]*Tensor['yz'])@b_amplitude,
    (F_hat[0]*Tensor['xz'] + F_hat[1]*Tensor['yz'] + F_hat[2]*Tensor['zz'])@b_amplitude
]).T
```

```python
del Tensor
gc.collect()
```

```python
ATA = A.T@A
ATd = A.T@observed_data[:,3]
```

```python
f0 = float(np.trace(ATA)/3)
```

```python
h_estimated = np.linalg.solve(ATA + 1e-5*f0*np.identity(3), ATd)
```

```python
h_estimated
```

```python
predicted_data2 = A@h_estimated
```

```python
del A
gc.collect()
```

```python
# Compute the residuals
residuals2 = predicted_data2 - observed_data[:,3]
```

```python
# Plot the data fit
fig = plt.figure(
    layout= 'tight', 
    figsize=(15,15)
)
mosaic = fig.subplot_mosaic(
    mosaic = '''
             abc
             ''',
    sharex=True, sharey=True
)

labels = [['(a)', '(b)', '(c)']]

mosaic_elements = [['a', 'b', 'c']]

for (mosaic_row, labels_row) in zip(
    mosaic_elements, 
    labels
):
    data_list = [
        predicted_data2, 
        observed_data[:,3],
        residuals2
    ]
    bounds_list = 2*[np.max(np.abs(observed_data[:,3]))] + [0.8*np.max(np.abs(residuals2))]
    titles = ['ACT calculada', 'ACT simulada', 'resíduos']
    
    for mosaic_element, data, bound, label, title in zip(mosaic_row, data_list, bounds_list, labels_row, titles):
    
        ax = mosaic[mosaic_element]
        ax.axis('scaled')
        im = ax.scatter(
            x=coordinates['y']*0.001, y=coordinates['x']*0.001, 
            c=data, s=10,
            cmap='seismic', vmin=-bound, vmax=bound
        )
        ax_divider = make_axes_locatable(ax)
        cax = ax_divider.append_axes("right", size="5%", pad="2%")
        cb = fig.colorbar(im, cax=cax)
        cb.ax.tick_params(labelsize=14)
        cb.ax.set_xlabel('nT', fontsize=16, loc='left')
        ax.set_ylim(0.001*study_area[0], 0.001*study_area[1])
        ax.set_xlim(0.001*study_area[2], 0.001*study_area[3])
        ax.tick_params(axis='x', labelsize=16)
        ax.tick_params(axis='y', labelsize=16)
        ax.grid()
        ax.annotate(label, xy=(0.02, 0.93), xycoords='axes fraction', fontsize=20)
        ax.annotate(title, xy=(0.02, 0.03), xycoords='axes fraction', fontsize=14)
    
for element in ['a']:
    mosaic[element].set_ylabel('x (km)', fontsize=18)
for element in ['a','b', 'c']:
    mosaic[element].set_xlabel('y (km)', fontsize=18)

plt.savefig('residuaos_camada_auxiliar.png', dpi=300)

plt.show()
```

### Estimate the second equivalent layer

```python
# Compute the unit vector defining the estimated total-magnetization direction
h_hat = utils.unit_vector(inc=inc_estimated, dec=dec_estimated)

# Compute the direction factors
a = utils.directional_factors(t=h_hat, u=F_hat)
```

```python
# compute the SEDM
R2 = inverse_distance.sedm(
    data_points = coordinates, 
    source_points = z_layer2
)
```

```python
# compute the second derivatives
Tensor = inverse_distance.grad_tensor(
    data_points = coordinates, 
    source_points = z_layer2,
    SEDM = R2,
    components=['xx', 'xy', 'xz', 'yy', 'yz', 'zz']
)
```

```python
# Compute the predicted data (approximated total-field anomaly)
A = np.zeros_like(Tensor['xx'])
for component in ['xx','xy','xz', 'yy', 'yz']:
    A += (a[component] * Tensor[component])
```

```python
del R2, Tensor
gc.collect()
```

```python
# estimate the equivalent layer
layer2_convergence, layer2 = eqlayer.method_CGLS(
    sensitivity_matrices=[A],
    data_vectors=[observed_data[:,3]],
    epsilon=1e-3,
    ITMAX=100,
    p0=None
)
```

```python
# Plot the estimated equivalent layer
bound = 1e-5*np.max(np.abs(layer2))

fig, ax = plt.subplots()

ax.axis('scaled')
im = ax.scatter(
        x=coordinates['y']*0.001, y=coordinates['x']*0.001, 
        c=1e-5*layer2, s=10,
        cmap='seismic', vmin=-bound, vmax=bound
    )
ax_divider = make_axes_locatable(ax)
cax = ax_divider.append_axes("right", size="5%", pad="2%")
cb = fig.colorbar(im, cax=cax)
cb.ax.tick_params(labelsize=14)
cb.ax.set_xlabel('$\\times 10^{5}$', fontsize=16, loc='left')
ax.set_ylim(0.001*study_area[0], 0.001*study_area[1])
ax.set_xlim(0.001*study_area[2], 0.001*study_area[3])
ax.tick_params(axis='x', labelsize=16)
ax.tick_params(axis='y', labelsize=16)
ax.grid()
ax.set_ylabel('x (km)', fontsize=18)
ax.set_xlabel('y (km)', fontsize=18)

plt.show()
```

```python
del fig, ax, im
gc.collect()
```

```python
# Convergence curve of the iterative total-field anomaly inversion
fig, ax = plt.subplots(figsize=(8,4))

iterations = [i for i in range(len(layer2_convergence))]
ax.plot(
    iterations, layer2_convergence, linestyle='-', c='k', alpha=1, markersize=4
)
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)
ax.set_xlabel('Iterations', fontsize=16)
ax.set_ylabel('Data fit', fontsize=16)
ax.grid(True)
plt.tight_layout()

plt.show()
```
